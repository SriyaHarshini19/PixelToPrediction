## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

Assumptions: Pages with declining performance but meaningful search visibility may have an opportunity for improvement through content updates.

Features: Recent impressions, CTR, search position, page age/update recency, and other historical performance signals available before the recommendation window.

Label definition: A page is considered a successful update candidate when it shows evidence of declining performance while still retaining meaningful search visibility.

Baseline: A simple hand-rule ranking using staleness × visibility × impressions is used as the baseline.

Validation design: Recommendations are evaluated on a later time window than the features used to create them, checking whether the ranking identifies pages that subsequently show improvement.

Leakage checks: Future performance windows, post-update information, product flags, and any fields created after the prediction point are excluded from features. This ensures recommendations use only information that would have been known when the decision was made.

## 5. Limitations

*What this work cannot claim.*

Limitation-What this work cannot claim
Observational: data	It cannot prove that updating a page will directly cause better search performance.
Historical: patterns	Past performance does not guarantee future results.
Ranking signals	The analysis cannot identify every factor affecting search rankings.
Data coverage:	Results only apply to the pages, metrics, and time windows available in the dataset.
Model performance:	A higher ranking score does not guarantee that every recommended page will improve.
Causality:	The model identifies promising pages to prioritize; it does not establish a causal relationship between an update and performance improvement.
